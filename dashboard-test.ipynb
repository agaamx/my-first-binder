1 CapStyle example:
#this plot is showing how by using different settings the lines ends can be shown 

```python
import matplotlib.pyplot as plt

plt.plot(
    [1, 2, 3],
    [1, 2, 3],
    linewidth=15,
    solid_capstyle='round'
)

plt.title("CapStyle Example")

plt.show()
```

Scatter Plot example:
#the scatter plot is very usrefull for showing correlation connecting differernt variables

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
```

Stem plot example:
#It is great for showing smaller datasets as in visualisation use dots and lines. It makes it very easy to understand the visualisation

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 2 * np.pi, 41)
y = np.exp(np.sin(x))

plt.stem(x, y)
plt.show()
```
